This workshop uses **CAMELS-GB-v2** — *Catchment Attributes and MEteorology for 
Large-sample Studies, Great Britain*.  
    
CAMELS-GB is an example of a **large-sample hydrological dataset** which contains data on **671 catchments** across Great Britain. Being able to examine hydrological behaviour across hundreds of catchments simultaneously is invaluable for hydrological process understanding and modelling, as it allows us to:
    
- Identify **patterns** in how catchments behave across different landscapes and climates
- Test whether **general relationships** hold across a wide range of conditions
- Build **predictive models** that can be applied to ungauged locations    

    
#### What data is provided in CAMELS-GB-v2?

| Data type | Coverage | Variables included |
|---|---|---|
| 🗺️ **Catchment boundaries** | 671 catchments | Polygons in British National Grid |
| 🏔️ **Catchment attributes** | Static summaries | Topography, climate, hydrology, soils, land cover, hydrogeology |
| 🌧️ **Meteorological time series** | 1970–2015, daily & hourly | Rainfall, PET, temperature, radiation, humidity |
| 💧 **River flow records** | 1970–2015, daily & hourly | Discharge, specific runoff |

## 1.1 Accessing catchment level summary data

Firstly, let's use  `geopandas`  to read in a shapefile containing the **Catchment Boundaries**:

````python
catchment_boundaries = gpd.read_file(home_dir + "Catchment_Boundaries/CAMELS_GB_v2_catchment_boundaries.shp")
catchment_boundaries.rename(columns={'ID':'gauge_id'}, inplace=True)
````

<!-- **Your task**   -->
And then, use  `pandas` to load **Catchment Attribute** data. These are csv files containing catchment level summaries of:  
- Climatic attributes
- Topographic attributes
- Hydrogeological attributes  
- Landcover attributes  
- Soil attributes  
- Hydrological signatures  

You can access these datasets using the file explorer at the left, and open them up. If you wish to work with, edit, or manipulate the datasets then you must load them into Python.

---

````python
hydrologic_attributes = pd.read_csv(home_dir + "Catchment_Attributes/CAMELS_GB_v2_hydrologic_attributes.csv")
hydrologic_attributes.head()
````

For explanations of the variables see _Table 5_ in this paper: https://essd.copernicus.org/preprints/essd-2025-608/essd-2025-608.pdf

````python
climatic_attributes = pd.read_csv(home_dir + "Catchment_Attributes/CAMELS_GB_v2_climatic_attributes.csv")
topography = pd.read_csv(home_dir + "Catchment_Attributes/CAMELS_GB_v2_topographic_attributes.csv")
hydrogeology = pd.read_csv(home_dir + "Catchment_Attributes/CAMELS_GB_v2_hydrogeology_attributes.csv")
landcover = pd.read_csv(home_dir + "Catchment_Attributes/CAMELS_GB_v2_landcover_attributes.csv")
soil = pd.read_csv(home_dir + 'Catchment_Attributes/CAMELS_GB_v2_soil_attributes.csv')
````

---

 We can then use the shared `gauge_id` column to join (or merge) all these datasets together, so we have all of the variables together in one place. 

---

````python
catchment_attributes = (topography
          .merge(climatic_attributes, on="gauge_id", how="left")
          .merge(hydrogeology, on="gauge_id", how="left")
          .merge(landcover, on="gauge_id", how="left")
          .merge(soil, on="gauge_id", how="left")
          .merge(hydrologic_attributes, on="gauge_id", how="left")
          .merge(catchment_boundaries[['gauge_id', 'geometry']], on="gauge_id", how="left"))
````

````python
catchment_attributes.head()
````

### 1.2 Plotting catchment summary data spatially
Now we have all the catchment attribute data in one place, we can make some maps which visualise how the catchment attributes vary spatially.


> 💡 Spatial visualisation is often the fastest way to spot structure in a 
> large dataset — patterns that would be invisible in a table of numbers become 
> immediately obvious on a map.

For this, we can use `geopandas` to convert our dataframe into a geodataframe which allows us to more easily work with location data. This uses the `geometry` column which was part of the catchment_boundaries file


---

````python
# Create geodataframe
catchment_attributes_gdf = gpd.GeoDataFrame(catchment_attributes, crs="EPSG:27700", geometry=catchment_attributes['geometry'])
# Convert the coordinate reference system (CRS), so as to work with basemaps
catchment_attributes_gdf = catchment_attributes_gdf.to_crs(epsg=3857)
````

---

Using this `geodataframe` we can then easily produce a map using `matplotlib` plotting functionalities.


Below, we create a map showing each catchment's mean daily precipitation.

---

````python
fig,ax=plt.subplots(figsize=(10,8))
catchment_attributes_gdf.plot(column = 'q_mean', # variable to plot
                              ax=ax, # where to plot it 
                              edgecolor='black',linewidth = 0.5, # specifying the edge colour and thickness
                              legend = True, # saying that I want a legend
                              cmap = 'Blues', #specifying the main colour scheme
                             norm=PowerNorm(gamma=0.8), #defining how to scale the colours
                             )
ax.set_title("Mean daily discharge (mm/day)")
ax.axis('off');
ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron, zoom=8)
````

### 📊 Introducing the Baseflow Index (BFI)

One variable in the CAMELS-GB dataset that captures catchment behaviour 
particularly well is the **Baseflow Index (BFI)** — the proportion of total 
streamflow sustained by groundwater, rather than direct surface runoff.

BFI captures something fundamental about a catchment's "personality". 
Consider two contrasting examples:

| | 🏔️ Upland Welsh catchment | 🌾 Chalk catchment (southern England) |
|---|---|---|
| **Geology** | Impermeable rock, thin soils | Permeable chalk aquifer |
| **Response to rainfall** | Rapid — water rushes straight to the river | Slow — rainfall infiltrates and is released gradually |
| **Flow after a storm** | Drops off quickly | Sustained for days to weeks |
| **BFI** | **Low** | **High** |

---

### 💧 Why does BFI matter?

| Application | Why BFI is relevant |
|---|---|
| 🌊 **Flood risk** | Low BFI catchments are flashy — water arrives fast and in large volumes |
| 🏜️ **Drought resilience** | High BFI catchments sustain flows through dry periods |
| 🐟 **River ecology** | Flow regime shapes the habitats and species a river can support |

---

I will start by testing the relationship of the baseflow index with the climate variables:

---

````python
# Specify the names of the climate variable columns
climatic_columns =  list(climatic_attributes.columns)
# Delete columns which are strings
cols_to_remove = climatic_attributes.select_dtypes(include="object").columns
climatic_columns = [x for x in climatic_columns if x not in cols_to_remove]
````

````python
# Compute correlations with the target (baseflow_index) [uses Pearson correlation coefficient]
corr_with_target = catchment_attributes[climatic_columns].corrwith(catchment_attributes['baseflow_index' ])

# Display as a sorted Series (absolute values)
corr_with_target.abs().sort_values(ascending=False)

# # Display as a sorted Series (not absolute)
corr_with_target.sort_values(ascending=False)
````

This shows us that the strongest relationships are a **positive relationship 
with mean PET** and a **negative relationship with mean daily precipitation**:

- **Higher rainfall → lower BFI**: Catchments with high precipitation tend 
to have lower baseflow contributions
- **Higher PET → higher BFI**: Catchments where more moisture is lost to 
evaporation tend to have higher baseflow contributions

> 🤔 **Think about it**: why might these relationships exist? What is it 
> about high-rainfall catchments that might lead to more surface runoff 
> and less groundwater contribution?

⚠️ NB: a statistical relationship doesn't always mean 
direct causation. Is PET *causing* high BFI, or are both simply 
characteristic of the same types of catchment?

<h1 style="color: white; margin: 0px;">🗺️ Research question 3. </h1>
<p style="margin: 5px 0px 0px 0px;">Can we predict catchment behaviour from landscape and climate attributes alone — even without flow data?</p>
</div>

In the previous sections, we **described** and **explained** spatial patterns 
in catchment behaviour, focusing on the baseflow index (the proportion of total  streamflow sustained by groundwater, rather than direct surface runoff.) We found that baseflow index varies substantially across GB (higher values in south east of England), and also saw that no single variable explains that variation cleanly — the controls are 
shaped by multiple factors simultaneously. 

This brings us to the next question:

> If we combine various catchment attributes together, can we **predict** the baseflow index
> for a catchment we've never measured?

---

### 🗺️ Why would we want to do this?

1. The baseflow index is very useful: The baseflow index is an important variable for understanding and predicting catchment response (in terms of flood events, drought events etc.)

2. Calculating the baseflow index requires flow measurements which are not available in lots of places: The majority of rivers worldwide have little or no flow record, yet decisions 
about flood risk, water resources, and infrastructure still need to be made 
for these locations. 

So, if we can learn the relationship between **catchment 
attributes** and **base flow index** from the catchments we *do* have data for, we can 
estimate behaviour anywhere we have landscape and climate data — even without 
a flow gauge.

---

### 🔨 What we'll build

We'll compare two models — a simple baseline and a more powerful alternative:

| Model | Type | Why? |
|---|---|---|
| **Linear Regression** | Simple, interpretable baseline | Assumes each predictor acts independently |
| **Random Forest** | Non-linear machine learning | Captures interactions between predictors |

## 3.3 Evaluting the linear model's performance

Sklearn also includes evaluation metrics which can be easily applied to evaluate the model's performance.  
E.g., R² and RMSE

The R² value tells us what proportion of variation in the baseflow index across catchments is explained by our chosen predictors (in this case, around 17%). 

The RMSE tells us the average magnitude of the model's prediction errors, expressed in the same units as the baseflow index. A lower RMSE indicates predictions that are closer to the observed values. Unlike R², which gives a relative measure of fit, RMSE gives a concrete sense of error size — for example, an RMSE of 0.12 means predictions are typically off by 0.12 BFI units.

Plotting a scatter plot of real values against predicted values also gives a visual indication of the model's performance.  

---

````python
r2 = r2_score(response_test, predictions)
rmse = np.sqrt(mean_squared_error(response_test, predictions))
print("Climate-only linear model performance:")
print("R2:", r2)
print("RMSE:", rmse)

# -------------------------
# Plot predicted vs observed 
# -------------------------
plt.scatter(predictions, response_test)
plt.xlabel('Predicted BFI')
plt.ylabel('Observed BFI')
# Add 1:1 line
plt.plot([response_test.min(), response_test.max()], [response_test.min(), response_test.max()], '--', color='red');
````

## 3.4 What did the linear model tell us?

Linear regression assumes that each predictor has a **straight-line, additive** 
relationship with the baseflow index — i.e. doubling rainfall doubles the effect on the baseflow index, 
regardless of soil type or topography. 

In reality, catchment behaviour is more complex than this. The effect of rainfall 
on runoff might depend heavily on soil type; steep, impermeable catchments might 
respond very differently to shallow, permeable ones even under the same rainfall. 
These kinds of **interactions and non-linearities** are difficult for a linear 
model to capture.

> 💡 This doesn't mean linear regression is useless — it's interpretable, fast, 
> and often a sensible starting point. But it sets a benchmark we can try to beat.

## 3.5 Can we do better? Introducing the Random Forest

To capture more complex relationships, we can turn to a **machine learning** approach 
called a **Random Forest**.

Before we build one, it helps to understand the building block: a **decision tree**.

### Decision trees

A decision tree makes predictions by asking a sequence of yes/no questions about the input variables. For instance, the decision tree below relates to the classification of an animal: 

----

![RF_Explained](InputFigures/DecisionTree_Explainer_Image.png)

---

The above example is for classification not regression. Regression works similarly but the outputs are numbers. 

In terms of predicting a continuous variable like the BFI, it will be similar to the above with each split chosen to best separate catchments with low vs high BFI. E.g., catchments with higher rainfall are likely to have higher BFI and those with lower rainfall are likely to have lower BFI. Similarly, catchments with low aridity, and so less evaporation, are also likely to have higher BFI. The tree keeps splitting until it reaches a stopping point, then predicts the average BFI of all catchments that fall into that final group.

---

![RF_Explained](InputFigures/DecisionTree_Explainer_Image2.png)

---

Decision trees are intuitive, but have a major weakness in that they tend to **overfit**. This means that they memorise the training data so well that they perform poorly on new, unseen catchments.

### From one (decision) tree to a (random) forest

A **Random Forest** fixes this by building **hundreds of trees**, each trained on:
- A **random subset of catchments** (so each tree sees different data)
- A **random subset of predictors** at each split (so trees don't all rely 
  on the same dominant variable)

The final prediction is the **average across all trees**. This averaging process — 
called **ensemble learning** — smooths out the quirks of individual trees and 
produces a much more robust model.

> 💡 Think of it like asking 500 different experts each with slightly different 
> experience, then averaging their answers — more reliable than any single expert.

![RF_Explained](InputFigures/RF_Explainer_Image.png)
Image from: https://thaddeus-segura.com/random-forest/  

### Why does this matter for hydrology?

Random Forests are well-suited to catchment prediction because:
- They handle **non-linear relationships** (e.g. the effect of slope on runoff 
  only matters above a certain threshold)
- They cope well with **correlated predictors** (e.g. rainfall and aridity 
  are related, but a RF can still use both sensibly)
- They produce **feature importance scores** that tell us which variables 
  the model found most useful — giving us hydrological insight, not just a prediction

## 3.6 Fitting the model

Okay, so let's now fit a random forest model. We'll use the same test and training data as we used for the linear model before. 

Is the performance any better than the linear model?

---

````python
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(predictor_train, response_train)
pred_rf = rf_model.predict(predictor_test)
````

````python
r2_rf = r2_score(response_test, pred_rf)
rmse_rf = np.sqrt(mean_squared_error(response_test, pred_rf))

print("Climate-only random forest model performance:")
print("R²:", round(r2_rf,3))
print("RMSE:", round(rmse_rf,3))

print("Climate-only linear model performance:")
print("R2:", r2)
print("RMSE:", rmse)

fig,axs =plt.subplots(ncols=2, figsize=(10,5))

axs[0].scatter(predictions, response_test)
axs[0].set_xlabel('Predicted BFI')
axs[0].set_ylabel('Observed BFI')
axs[0].set_title("Linear model")
# Add 1:1 line
axs[0].plot([response_test.min(), response_test.max()], [response_test.min(), response_test.max()], '--', color='red');

axs[1].scatter(pred_rf, response_test)
axs[1].set_xlabel('Predicted BFI')
axs[1].set_ylabel('Observed BFI')
axs[1].set_title("Random Forest")
# Add 1:1 line
axs[1].plot([response_test.min(), response_test.max()], [response_test.min(), response_test.max()], '--', color='red');
````
